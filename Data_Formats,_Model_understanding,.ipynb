```python
import os
import pandas as pd

files_to_compare = {
    "CSV": file_name,
    "Excel": excel_file,
    "JSON": json_file,
    "Parquet": parquet_file
}

size_results = []

for format_name, path in files_to_compare.items():
    size_kb = os.path.getsize(path) / 1024
    size_results.append({
        "Format": format_name,
        "Size (KB)": round(size_kb, 2)
    })

size_df = pd.DataFrame(size_results)

display(size_df.sort_values("Size (KB)"))
```

```python
import time

load_times = []

for format_name, path in files_to_compare.items():

    start_time = time.perf_counter()

    if format_name == "CSV":
        temp_df = pd.read_csv(path)
    elif format_name == "Excel":
        temp_df = pd.read_excel(path)
    elif format_name == "JSON":
        temp_df = pd.read_json(path, orient="records")
    elif format_name == "Parquet":
        temp_df = pd.read_parquet(path)

    elapsed_time = time.perf_counter() - start_time

    load_times.append({
        "Format": format_name,
        "Load time (seconds)": round(elapsed_time, 5)
    })

load_time_df = pd.DataFrame(load_times)

display(load_time_df.sort_values("Load time (seconds)"))

# "In this block, I am measuring the time required to load the same dataset from four different file formats.
# I use Python's high-resolution performance timer to record the start and finish times of each read operation.
# Depending on the format, I call the appropriate Pandas reader. I calculate the elapsed time, store the result, and
# sort the results from fastest to slowest. This gives me an initial comparison of loading performance. However,
# I should repeat the test under controlled conditions before drawing a strong conclusion."
```

```python
# Clean TotalCharges


total_charges_numeric = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

print("Original TotalCharges data type:", df["TotalCharges"].dtype)
print("Missing or non-numeric values:", total_charges_numeric.isna().sum())

# In this block, I am investigating the TotalCharges column before cleaning it.

# The `pd.to_numeric()` function attempts to convert the column into numerical values.

# I use `errors="coerce"` so that any value that cannot be converted becomes a missing value,
# represented by NaN, instead of causing the code to fail.

# Finally, I print the original data type and count the missing or non-numeric values.
# This helps me understand the scale of the data-quality issue before deciding how to handle it.

# Notice that I have not modified the original dataset yet. I am first checking the problem
# so that the next cleaning step is based on evidence.
```

```python
# Inspect the affected records


invalid_total_charges = df.loc[
    total_charges_numeric.isna(),
    ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]
]

display(invalid_total_charges)

# In this block, I am inspecting the records where TotalCharges could not be converted into a number.

# The `isna()` method identifies the invalid values in the converted column.

# The `loc` accessor selects only those records and displays five relevant columns:
# customer ID, tenure, monthly charges, total charges, and churn status.

# By comparing tenure and monthly charges with TotalCharges,
# I can investigate whether these are genuinely missing values or whether there is a reasonable explanation for them.

# This helps me make a data-cleaning decision based on evidence rather than assumptions.
```

```python
# Clean TotalCharges


df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
).fillna(0)

print("TotalCharges data type:", df["TotalCharges"].dtype)
print("Missing TotalCharges values:", df["TotalCharges"].isna().sum())
print("Records with zero tenure:", (df["tenure"] == 0).sum())

# In this block, I am cleaning the TotalCharges column.

# First, I convert the column into a numerical data type using `pd.to_numeric()`.

# The `errors="coerce"` parameter converts invalid values into missing values.

# Next, `fillna(0)` replaces missing values with zero. Based on our inspection,
# all 11 affected customers have zero tenure, so zero total charges is a reasonable interpretation for this dataset.

# Finally, I verify the data type, check whether any missing values remain, and count customers with zero tenure.

# This step prepares TotalCharges for machine learning while making the cleaning decision transparent and verifiable.
```

```python
#  Prepare features and target


X = df.drop(columns=["Churn", "customerID"])
y = df["Churn"].map({"No": 0, "Yes": 1})

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget values:")
print(y.value_counts())

# In this block, I am separating the input features from the target variable.

# The variable X contains the customer information that the model will use to make predictions. I remove Churn because it is the answer we want the model to learn to predict. I also remove customerID because it is an identifier rather than a meaningful behavioral feature.

# The variable y contains the actual churn outcome.
#  I convert No to zero and Yes to one because machine-learning algorithms work conveniently with numerical target labels.

# Finally, I print the shapes of X and y and check the target counts to confirm that the features and labels are aligned.

# This separation is the foundation of supervised machine learning.
```

```python
# Identify numerical and categorical columns


numerical_columns = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_columns = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numerical columns:", len(numerical_columns))
print(numerical_columns)

print("\nCategorical columns:", len(categorical_columns))
print(categorical_columns)

# Before splitting the data, I am identifying which features are numerical and which are categorical.

# I use `select_dtypes()` to select columns based on their data types.

# Numerical columns contain values such as tenure and monthly charges.
# Categorical columns contain labels such as contract type, payment method, and internet service.

# This distinction matters because different types of features require different preprocessing methods.

# For this project, we will keep numerical features as numerical values and
# apply one-hot encoding to categorical features. We will then build a preprocessing pipeline and
# split the data before fitting any transformations, so information from the test set cannot leak into training.
```

```python
# Split the data before preprocessing
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining churn rate:", round(y_train.mean() * 100, 2), "%")
print("Testing churn rate:", round(y_test.mean() * 100, 2), "%")

```

```python
# Build the preprocessing pipeline


from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numerical_columns),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_columns)
    ]
)

print("Preprocessing pipeline created successfully.")
print("Numerical columns:", len(numerical_columns))
print("Categorical columns:", len(categorical_columns))

# In this block, I am building a preprocessing pipeline for my customer churn model.

# I use ColumnTransformer to apply different preprocessing rules to different groups of columns.

# For numerical columns, I use passthrough, which keeps their values unchanged.

# For categorical columns, I use OneHotEncoder to convert categories into numerical indicator columns.
# I set handle_unknown to ignore so the pipeline can process categories that were not present when the encoder was fitted.

# I also import Pipeline because we will use it to connect preprocessing with the machine-learning model in a later step.

# At this stage, I am only defining the preprocessing instructions. I am not fitting the encoder or training the model yet.
#  We will fit everything using the training data only.
```

```python
# Fit and transform the training data


X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Training data before preprocessing:", X_train.shape)
print("Training data after preprocessing:", X_train_processed.shape)

print("\nTesting data before preprocessing:", X_test.shape)
print("Testing data after preprocessing:", X_test_processed.shape)


# In this block, I am applying the preprocessing instructions to the training and testing datasets.

# For the training data, I use fit_transform.
# The fit operation learns the categories present in the training set, and the transform operation converts the features into the required numerical representation.

# For the testing data, I use only transform.
# This reuses the categories learned from the training data instead of learning anything new from the test set.

# I then compare the shapes before and after preprocessing.
# The number of rows should remain unchanged,
# but the number of columns will usually increase because one-hot encoding creates additional columns for categorical values.

# This step prepares the features for machine learning while keeping the test data separate from the fitting process.
```

```python
# Check feature cardinality


feature_summary = pd.DataFrame({
    "Data_Type": X.dtypes.astype(str),
    "Unique_Values": X.nunique(),
    "Missing_Values": X.isna().sum()
})

display(
    feature_summary.sort_values(
        "Unique_Values",
        ascending=False
    )
)

# Before training the model, I am inspecting the remaining features to understand their data types, number of distinct values,
#  and missing values.

# The `nunique()` method counts the distinct values in each column.
# This helps identify features with many unique values, which may need special attention, and features with very few values, such as binary categories.

# The missing-value count helps us identify any remaining data-quality issues.

# I will use this summary to check whether any features should be excluded or require additional preprocessing.
# I won't automatically remove a feature just because it has few unique values or seems less important.
#  Feature selection should be justified by the data and evaluated against model performance.
```

```python
# Train the baseline Logistic Regression model


from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

model.fit(X_train_processed, y_train)

print("Baseline Logistic Regression trained.")
print("Training samples:", X_train_processed.shape[0])
print("Features used:", X_train_processed.shape[1])

# I have inspected the feature summary and found no remaining missing values in the input features.
#  I also confirmed that the remaining columns contain meaningful customer information rather than an obvious identifier.

# I will now train a baseline Logistic Regression model using all 45 processed features.

# A baseline gives me a reference point.
# Later, I can examine the model coefficients and evaluate whether removing selected features improves performance.

# The fit method learns the relationship between the training features and the churn labels.
# Finally, I print the number of training samples and processed features to verify the training setup.
```

```python
# Update the preprocessing pipeline


from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_columns),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_columns)
    ]
)

print("Preprocessing updated with numerical feature scaling.")

# The model training produced a convergence warning, so I am improving the preprocessing before continuing.

# Previously, I passed numerical features through unchanged.
# Now, I am using StandardScaler to standardize them.
# This transforms numerical values to a common scale based on the training data's mean and standard deviation when the pipeline is fitted.

# Scaling often helps Logistic Regression's optimization algorithm converge more reliably,
# especially when features have very different magnitudes.

# I am keeping one-hot encoding for categorical features and
# setting handle_unknown to ignore so unseen categories can be handled safely.

# This block only defines the updated preprocessing instructions.
# In the next step, I will fit the updated preprocessing on the training data and transform both datasets consistently.
```

```python
# Reprocess training and testing data


X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Training data shape:", X_train_processed.shape)
print("Testing data shape:", X_test_processed.shape)
print("Preprocessing completed.")


# In this block, I am applying the updated preprocessing pipeline.

# I use fit_transform on the training data.
# This learns the scaling parameters from the training numerical features and learns the categorical values for one-hot encoding.

# I then use transform on the testing data, applying the same learned scaling parameters and category mapping without fitting again.

# Finally, I print both shapes to verify that preprocessing completed successfully.
# The number of rows should remain unchanged, and both datasets should have the same number of processed features.

# This ensures that the model will train and be evaluated using consistently transformed data,
# without learning preprocessing information from the test set.
```

```python
# Retrain Logistic Regression


from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

model.fit(X_train_processed, y_train)

print("Logistic Regression retrained successfully.")
print("Iterations used:", model.n_iter_[0])

# In this block, I am retraining Logistic Regression using the updated,
# standardized numerical features and one-hot encoded categorical features.

# I set the maximum number of iterations to 2,000 to give the optimization algorithm more opportunity to converge.

# The fit method learns the model coefficients from the training data and its churn labels.

# Finally, I print the number of iterations the model actually used.
# This helps me inspect how the optimization process progressed.
# If a convergence warning still appears, we will investigate further rather than assume that increasing the iteration limit fixed it.
```

```python
# Make predictions on test data


y_pred = model.predict(X_test_processed)

print("Number of test customers:", len(y_test))
print("Number of predictions:", len(y_pred))

print("\nFirst 10 actual churn labels:")
print(y_test.iloc[:10].tolist())

print("\nFirst 10 predicted churn labels:")
print(y_pred[:10].tolist())

# In this block, I am using the trained model to predict churn for the testing dataset.

# The predict method takes the processed test features and returns a predicted class for each customer.
# Zero means the model predicts that the customer will stay, and one means it predicts that the customer will churn.

# I print the number of actual test labels and predictions to verify that every test customer has a corresponding prediction.

# Then I compare the first ten actual labels with the first ten predicted labels.
# This gives me an initial look at the predictions, although we need formal performance metrics to evaluate the model reliably.
```

The model generated 1,409 predictions, matching the number of customers in the test set.

When I compare the first ten actual labels with the predicted labels, I see that eight predictions are correct and two are incorrect.

The model incorrectly predicts churn for two customers who actually stayed, and it predicts that one customer will stay when that customer actually churned.

This is only a small sample, so I cannot use it to judge the model's overall performance. I need metrics such as accuracy, precision, recall, and F1-score to understand how well it performs across the complete test dataset.

```python
# Evaluate accuracy


from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Test accuracy:", round(accuracy * 100, 2), "%")

# In this block, I am calculating the model's accuracy on the test dataset.

# Accuracy is the number of correct predictions divided by the total number of predictions.

# I use `accuracy_score()` to compare the actual churn labels with the model's predicted labels,
# then multiply the result by 100 to express it as a percentage.

# Accuracy gives me a useful first measure of performance.
#  However, our dataset is imbalanced because only about 26.54 percent of customers churned.
#  Therefore, I will also examine precision, recall, and F1-score to understand
#  whether the model identifies customers who are actually at risk of leaving.
```

```python
# Calculate precision, recall, and F1-score


from sklearn.metrics import precision_score, recall_score, f1_score

precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Precision:", round(precision * 100, 2), "%")
print("Recall:", round(recall * 100, 2), "%")
print("F1-score:", round(f1 * 100, 2), "%")

# In this block, I am calculating three important classification metrics.

# Precision measures, among the customers the model predicted would churn, how many actually churned.

# Recall measures, among all customers who actually churned, how many the model successfully identified.

# F1-score combines precision and recall into a single metric using their harmonic mean. It is useful when I want to balance the two.

# I am calculating these metrics using the test labels and predictions, with churn represented by one as the positive class.

# These metrics will help me understand whether the model's accuracy translates into useful churn detection for the business.
```

The model has generated a churn probability for each customer in the test set.

For example, the first customer's estimated churn probability is 4.6 percent, while the second customer's probability is 68.4 percent.

The model's probabilities range from 0.2 percent to 85.4 percent in this test set.

At the default threshold of 0.5, probabilities at or above 50 percent are classified as churn, and probabilities below 50 percent are classified as no churn.

The important point is that the model produces probabilities first, and the classification threshold converts those probabilities into final predictions. By adjusting that threshold, we can change the balance between identifying churners and avoiding false alarms.

```python
# Compare different thresholds


from sklearn.metrics import precision_score, recall_score, f1_score

for threshold in [0.3, 0.4, 0.5]:
    y_threshold_pred = (y_prob >= threshold).astype(int)

    print(f"\nThreshold: {threshold}")
    print("Precision:", round(precision_score(y_test, y_threshold_pred) * 100, 2), "%")
    print("Recall:", round(recall_score(y_test, y_threshold_pred) * 100, 2), "%")
    print("F1-score:", round(f1_score(y_test, y_threshold_pred) * 100, 2), "%")
```

The threshold comparison shows that changing the classification threshold changes the balance between precision and recall.

At the default threshold of 0.5, precision is 65.72 percent and recall is 55.88 percent.

When I lower the threshold to 0.3, recall increases to 75.4 percent. This means the model identifies a larger proportion of actual churners. However, precision falls to 51.93 percent, so more customers flagged as churn risks may actually stay.

The threshold of 0.4 provides an intermediate result, with 56.82 percent precision and 66.84 percent recall.

If the business prioritizes identifying more customers before they leave, I would investigate the lower threshold. Before making a final decision, I would also consider retention costs and validate the chosen threshold on separate validation data.

```python
# Calculate ROC-AUC


from sklearn.metrics import roc_auc_score

roc_auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC score:", round(roc_auc, 4))

# In this block, I am calculating the ROC-AUC score for the churn prediction model.

# Unlike accuracy, precision, and recall at a specific threshold,
# ROC-AUC evaluates how well the model ranks customers who actually churned above customers
# who stayed across different classification thresholds.

# A score closer to one indicates stronger separation between the two groups,
#  while a score around 0.5 indicates performance similar to random ranking.

# I use the actual test labels and the predicted churn probabilities, rather than the binary predictions,
# because ROC-AUC evaluates ranking across thresholds.

# This gives me another perspective on model performance that is not tied to choosing a single threshold.
```

My Logistic Regression model achieved a ROC-AUC score of 0.8421 on the test dataset.

ROC-AUC measures the model's ability to distinguish between customers who churn and customers who stay across different classification thresholds.

A score of approximately 0.84 indicates that the model has useful ranking ability. For a randomly selected churner and non-churner, it will rank the churner higher approximately 84 percent of the time, subject to the handling of tied scores.

This is different from accuracy. Accuracy measures the proportion of correct predictions at a particular threshold, while ROC-AUC evaluates ranking performance across thresholds.

The result suggests that the model can distinguish between churn risks reasonably well, although we still need to consider recall, precision, and business costs when choosing how to use it.
